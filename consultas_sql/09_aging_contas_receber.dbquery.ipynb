```sql
SELECT
    faixa_aging,

    quantidade_titulos,
    quantidade_clientes,

    valor_carteira,
    participacao_carteira_percentual

FROM aging_contas_receber

ORDER BY

    CASE faixa_aging

        WHEN 'A vencer'
            THEN 1

        WHEN 'Vencido de 1 a 30 dias'
            THEN 2

        WHEN 'Vencido de 31 a 60 dias'
            THEN 3

        WHEN 'Vencido de 61 a 90 dias'
            THEN 4

        WHEN 'Vencido acima de 90 dias'
            THEN 5

        ELSE 6

    END;
```
