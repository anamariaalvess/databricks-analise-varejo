```sql
SELECT
    mes_referencia,
    ano_mes,

    faturamento,
    faturamento_bruto,
    valor_desconto,

    custo_total,
    lucro_bruto,
    margem_percentual,

    quantidade_pedidos,
    clientes_ativos,
    produtos_vendidos,
    quantidade_vendida,

    ticket_medio,

    crescimento_faturamento_percentual

FROM vendas_mensais

ORDER BY mes_referencia;
```
