```sql
WITH ultimo_mes AS (

    SELECT
        MAX(mes_referencia) AS mes_referencia

    FROM desempenho_vendedores_mensal
)

SELECT
    v.mes_referencia,

    v.ranking_faturamento,

    v.id_vendedor,
    v.nome_vendedor,

    v.regiao,
    v.nivel,

    v.faturamento,
    v.valor_meta,

    v.atingimento_meta_percentual,

    v.lucro_bruto,
    v.margem_percentual,

    v.quantidade_pedidos,
    v.clientes_positivados,

    v.ticket_medio,

    v.status_meta

FROM desempenho_vendedores_mensal v

INNER JOIN ultimo_mes u
    ON v.mes_referencia = u.mes_referencia

ORDER BY
    v.ranking_faturamento;
```
