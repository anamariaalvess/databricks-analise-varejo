```sql
WITH ultimo_mes AS (

    SELECT
        mes_referencia,
        faturamento,
        faturamento_mes_anterior,
        lucro_bruto,
        margem_percentual,
        quantidade_pedidos,
        clientes_ativos,
        produtos_vendidos,
        ticket_medio,
        percentual_desconto,
        crescimento_faturamento_percentual,

        ROW_NUMBER() OVER (
            ORDER BY mes_referencia DESC
        ) AS ordem

    FROM vendas_mensais
)

SELECT
    mes_referencia,

    faturamento,
    faturamento_mes_anterior,

    lucro_bruto,
    margem_percentual,

    quantidade_pedidos,
    clientes_ativos,
    produtos_vendidos,

    ticket_medio,
    percentual_desconto,

    crescimento_faturamento_percentual

FROM ultimo_mes

WHERE ordem = 1;
```
