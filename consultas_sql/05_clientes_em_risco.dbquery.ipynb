```sql
SELECT
    id_cliente,
    nome_cliente,

    segmento_cliente,
    porte_cliente,

    cidade,
    estado,

    data_ultima_compra,

    recencia_dias,
    frequencia,

    valor_monetario,
    ticket_medio,

    score_r,
    score_f,
    score_m,

    segmento_rfm,
    prioridade_acao

FROM clientes_rfm

WHERE segmento_rfm IN (
    'Grandes clientes em risco',
    'Em risco'
)

ORDER BY
    valor_monetario DESC;
```
