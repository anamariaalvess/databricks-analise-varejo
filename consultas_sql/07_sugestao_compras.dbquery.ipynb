```sql
SELECT
    id_produto,
    nome_produto,
    categoria,

    classe_abc,
    ranking_faturamento,

    venda_media_diaria,

    quantidade_estoque,
    quantidade_pendente_recebimento,

    estoque_projetado,
    ponto_reposicao,
    estoque_alvo,

    quantidade_sugerida_compra,

    nome_fornecedor,
    prazo_medio_entrega,
    avaliacao_fornecedor,

    custo_unitario,
    valor_estimado_compra,

    prioridade_compra,
    recomendacao

FROM sugestao_compras

WHERE quantidade_sugerida_compra > 0

ORDER BY

    CASE prioridade_compra

        WHEN 'Crítica'
            THEN 1

        WHEN 'Muito alta'
            THEN 2

        WHEN 'Alta'
            THEN 3

        WHEN 'Normal'
            THEN 4

        ELSE 5

    END,

    ranking_faturamento;
```
