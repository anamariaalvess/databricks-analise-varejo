# Configuração do ambiente

## Objetivo

Este notebook realiza a configuração inicial do projeto **Análise de Varejo com Databricks**.

O projeto simula o ambiente analítico de uma distribuidora fictícia, **Distribuidora Horizonte**, e utiliza uma arquitetura Lakehouse baseada no padrão Medallion.

## Arquitetura

Os dados serão organizados nas seguintes camadas:

- **Landing:** armazenamento dos arquivos fictícios antes da ingestão.
- **Bronze:** dados brutos, preservando a estrutura de origem.
- **Silver:** dados tratados, padronizados e integrados.
- **Gold:** dados preparados para análises, indicadores e dashboards.

## Tecnologias

- Databricks
- Apache Spark
- PySpark
- Delta Lake
- Unity Catalog
- Databricks SQL
- AI/BI Dashboards
- Genie

## Observação

Todos os dados utilizados neste projeto são fictícios e foram gerados exclusivamente para fins educacionais e de portfólio.

```python
# Identifica o catálogo e o schema atualmente utilizados pelo workspace

catalogo_atual = spark.sql("SELECT current_catalog()").first()[0]
schema_atual = spark.sql("SELECT current_schema()").first()[0]

print(f"Catálogo atual: {catalogo_atual}")
print(f"Schema atual: {schema_atual}")
```

```python
# Define os schemas utilizados pelo projeto

schemas_projeto = [
    "varejo_landing",
    "varejo_bronze",
    "varejo_silver",
    "varejo_gold"
]

# Cria os schemas

for schema in schemas_projeto:
    spark.sql(
        f"""
        CREATE SCHEMA IF NOT EXISTS `{catalogo_atual}`.`{schema}`
        """
    )

    print(f"Schema criado ou já existente: {catalogo_atual}.{schema}")
```

```python
# Adiciona descrições aos schemas para facilitar a identificação no Unity Catalog

comentarios_schemas = {
    "varejo_landing": "Área de armazenamento dos arquivos fictícios antes da ingestão para a camada Bronze.",
    "varejo_bronze": "Camada Bronze contendo os dados brutos ingeridos das fontes fictícias.",
    "varejo_silver": "Camada Silver contendo dados tratados, padronizados, validados e integrados.",
    "varejo_gold": "Camada Gold contendo dados analíticos, indicadores e tabelas utilizadas pelos dashboards."
}

for schema, comentario in comentarios_schemas.items():

    spark.sql(
        f"""
        COMMENT ON SCHEMA `{catalogo_atual}`.`{schema}`
        IS '{comentario}'
        """
    )

print("Comentários adicionados aos schemas.")
```

```python
# Cria o Volume que armazenará os arquivos fictícios do projeto

spark.sql(
    f"""
    CREATE VOLUME IF NOT EXISTS
    `{catalogo_atual}`.`varejo_landing`.`dados_brutos`
    COMMENT 'Arquivos fictícios utilizados como fontes de dados do projeto de análise de varejo'
    """
)

print("Volume dados_brutos criado com sucesso.")
```

```python
# Define o caminho principal utilizado para armazenamento dos arquivos fictícios

caminho_volume = (
    f"/Volumes/{catalogo_atual}/varejo_landing/dados_brutos"
)

print(f"Caminho do Volume: {caminho_volume}")
```

```python
# Define as fontes de dados que serão simuladas no projeto

fontes = [
    "clientes",
    "produtos",
    "vendedores",
    "fornecedores",
    "vendas",
    "compras",
    "estoque",
    "contas_receber",
    "contas_pagar",
    "metas_vendas"
]

# Cria uma pasta para cada fonte dentro do Volume

for fonte in fontes:
    dbutils.fs.mkdirs(f"{caminho_volume}/{fonte}")

print("Estrutura de diretórios criada com sucesso.")
```

```python
# Exibe os diretórios existentes no Volume para validar a configuração

display(
    dbutils.fs.ls(caminho_volume)
)
```
